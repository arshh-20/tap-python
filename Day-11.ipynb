##### 4. Read olist_orders_dataset.csv with the csv module and print the row count.

```python
import csv

with open("data/olist_orders_dataset.csv", "r", newline="") as file:
    reader = csv.reader(file)
    next(reader) 

    row_count = sum(1 for row in reader)

print("Row count:", row_count)
```

##### 7. Loop over all Olist CSV files and print a summary for each.

```python
files = ['data/olist_customers_dataset.csv', 
         'data/olist_order_payments_dataset.csv', 
         'data/olist_products_dataset.csv', 
         'data/olist_geolocation_dataset.csv', 
         'data/olist_order_reviews_dataset.csv', 
         'data/olist_sellers_dataset.csv', 
         'data/olist_order_items_dataset.csv', 
         'data/olist_orders_dataset.csv', 
         'data/product_category_name_translation.csv'
         ]
```

```python
for file in files: 
    try: 
        df = pd.read_csv("data/olist_customers_dataset.csv")
        print(f"Summary of {file}")
        df.info()
    except FileNotFoundError:
        print(f"File: {file} not found")
```
